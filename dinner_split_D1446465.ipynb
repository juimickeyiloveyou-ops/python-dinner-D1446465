```python
# ==========================================
# Python 聚餐費用試算器 - 三組測試實作
# ==========================================

# 共用基礎設定：服務費率與預算
service_rate = 0.10
budget_per_person = 300

# ------------------------------------------
# 測試一：使用原始資料（人數 = 4, 折價券 = 100）
# ------------------------------------------
print("=== 測試一 ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
discount = 100

# 1. 整理聚餐名稱：去除頭尾空白並轉成小寫
cleaned_event_name = event_name.strip().lower()

# 2. 整理餐費與人數並轉換型別
# 移除 "NT$", ",", 與空白
cleaned_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
cleaned_meal_amount = float(cleaned_meal_str) # 轉成數值供計算
cleaned_people = int(people)                  # 文字人數轉成整數

# 3. 分步計算費用
service_fee = cleaned_meal_amount * service_rate              # 服務費：原始餐費 × 服務費率
total_amount = cleaned_meal_amount + service_fee - discount   # 應付總額：原始餐費 + 服務費 - 折價券
cost_per_person = total_amount / cleaned_people               # 每人費用：應付總額 ÷ 分攤人數

# 4. 判斷是否超過預算（布林變數）
is_over_budget = cost_per_person > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{cleaned_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{cleaned_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


# ------------------------------------------
# 測試二：只將人數改成 "5"，折價券維持 100 元
# ------------------------------------------
print("=== 測試二 ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "5"  # 修改人數
discount = 100

cleaned_event_name = event_name.strip().lower()
cleaned_meal_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
cleaned_people = int(people)

service_fee = cleaned_meal_amount * service_rate
total_amount = cleaned_meal_amount + service_fee - discount
cost_per_person = total_amount / cleaned_people
is_over_budget = cost_per_person > budget_per_person

print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{cleaned_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{cleaned_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


# ------------------------------------------
# 測試三：人數恢復為 "4"，只將折價券改成 220 元
# ------------------------------------------
print("=== 測試三 ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
discount = 220  # 修改折價券

cleaned_event_name = event_name.strip().lower()
cleaned_meal_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
cleaned_people = int(people)

service_fee = cleaned_meal_amount * service_rate
total_amount = cleaned_meal_amount + service_fee - discount
cost_per_person = total_amount / cleaned_people
is_over_budget = cost_per_person > budget_per_person

print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{cleaned_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{cleaned_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
